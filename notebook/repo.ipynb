```python
import re
import html


def clean_text(text):
    
    # Convert anything that isn't text into an empty string
    text = str(text)

    # Decode HTML entities
    text = html.unescape(text)

    # Remove HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Remove Markdown images
    text = re.sub(r"!\[.*?\]\(.*?\)", " ", text)

    # Remove Markdown links but keep the link text
    text = re.sub(r"\[(.*?)\]\(.*?\)", r"\1", text)

    # Remove excessive whitespace
    text = re.sub(r"\s+", " ", text)

    # Remove leading/trailing spaces
    text = text.strip()

    return text
```

```python
df["clean_text"] = df["combined_text"].apply(clean_text)
```

```python
for idx in top_indices:
    idx = idx.item()

    print("Repository:", df.iloc[idx]["name"])
    print("Description:", df.iloc[idx]["description"])
    print("Language:", df.iloc[idx]["language"])
    print("Similarity:", round(similarities[idx].item(), 4))
    print("-" * 60)
```

```python
def search_repositories(query, top_k=5):

    # Convert user query into embedding
    query_embedding = model.encode(query)

    # Compare query with all repository embeddings
    similarities = util.cos_sim(query_embedding, embeddings)[0]

    # Get top K repository indexes
    top_indices = similarities.argsort(descending=True)[:top_k]

    results = []

    for idx in top_indices:

        idx = idx.item()

        results.append({
            "name": df.iloc[idx]["name"],
            "description": df.iloc[idx]["description"],
            "language": df.iloc[idx]["language"],
            "stars": df.iloc[idx]["stars"],
            "url": df.iloc[idx]["url"],
            "similarity": round(similarities[idx].item(), 4)
        })

    return results
```

```python
results = search_repositories(
    "Find me a Python deep-learning project for image classification."
)

results
```

```python
rea = search_repositories("Python NLP sentiment analysis")
rea
```
